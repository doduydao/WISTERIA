```python
# Định nghĩa luật chuyển đổi label
transitivity_rules = {
    ("BEFORE", "BEFORE"): "BEFORE",
    ("AFTER", "AFTER"): "AFTER"
    # ("BEFORE", "OVERLAP"):"BEFORE_OVERLAP"
}

symmetry_rules = {
    "BEFORE": "AFTER",
    "AFTER": "BEFORE",
    "OVERLAP": "OVERLAP",
    "SIMULTANEOUS": "SIMULTANEOUS"
    # "BEGUN_BY": "ENDED_BY",
    # "ENDED_BY": "BEGUN_BY",
    # "DURING": "DURING",
}

def add_transitivity(df, transitivity_rules, label_mapping, seen_rows):
    new_rows = []
    
    # Chuyển đổi seen_rows thành một set có định dạng tương tự valid_pairs["new_row"]
    seen_rows_set = set(f"{row[0]}_{row[1]}_{row[2]}_{row[3]}" for row in seen_rows)

    # Nhóm dữ liệu theo 'document_id'
    grouped = df.groupby('document_id')

    for _, group in grouped:
        # Thêm cột tạm để dễ xử lý
        group = group.assign(pair_key=group["entity1_id"].astype(str) + "_" + group["entity2_id"].astype(str))

        # Merge để tìm các cặp hàng thỏa mãn entity2_id của hàng trước = entity1_id của hàng sau
        merged = pd.merge(
            group, group,
            left_on="entity2_id", right_on="entity1_id",
            suffixes=("_1", "_2")
        )

        # Lọc các cặp hàng thỏa mãn quy tắc chuyển tiếp
        merged["label_pair"] = list(zip(merged["label_1"], merged["label_2"]))
        valid_pairs = merged[merged["label_pair"].isin(transitivity_rules)]

        # Tạo các hàng mới dựa trên quy tắc chuyển tiếp
        valid_pairs["new_label"] = valid_pairs["label_pair"].map(transitivity_rules)
        # valid_pairs["new_row"] = (
        #     valid_pairs["entity1_id_1"].astype(str) + "_" +
        #     valid_pairs["entity2_id_2"].astype(str) + "_" +
        #     valid_pairs["document_id_1"].astype(str) + "_" +
        #     valid_pairs["new_label"]
        # )

        # Tạo hàng mới dưới dạng tuple để kiểm tra trùng lặp
        valid_pairs["new_row"] = valid_pairs.apply(
            lambda row: (
                row["entity1_id_1"], row["entity2_id_2"], row["document_id_1"], row["new_label"]
            ),
            axis=1
        )
        
        # Lọc các hàng chưa xuất hiện trong seen_rows
        unique_valid_pairs = valid_pairs[~valid_pairs["new_row"].isin(seen_rows)]
 
        # Thêm các hàng mới vào seen_rows
        # seen_rows.update(valid_pairs[["entity1_id_1", "entity2_id_2", "document_id_1", "new_label"]].apply(
        #     lambda row: (row["entity1_id_1"], row["entity2_id_2"],  row["document_id_1"],  row["new_label"]),
        #     axis=1
        # ))
        seen_rows.update(unique_valid_pairs["new_row"].tolist())
        
        # Tạo DataFrame các hàng mới
        new_rows_group = unique_valid_pairs[[
            "entity1_id_1", "entity2_id_2",
            "entity1_start_1", "entity2_start_2",
            "entity1_end_1", "entity2_end_2",
            "entity1_text_1", "entity2_text_2",
            "document_id_1", "text_1", "new_label"
        ]].rename(columns={
            "entity1_id_1": "entity1_id",
            "entity2_id_2": "entity2_id",
            "entity1_start_1": "entity1_start",
            "entity2_start_2": "entity2_start",
            "entity1_end_1": "entity1_end",
            "entity2_end_2": "entity2_end",
            "entity1_text_1": "entity1_text",
            "entity2_text_2": "entity2_text",
            "document_id_1": "document_id",
            "text_1": "text",
            "new_label": "label"
        })

        # Thêm cột label_encoded vào các hàng mới
        new_rows_group["label_encoded"] = new_rows_group["label"].map(label_mapping)

        # Thêm các hàng mới vào danh sách tổng
        new_rows.append(new_rows_group)

    # Kết hợp các nhóm lại và trả về DataFrame mới
    new_rows_df = pd.concat(new_rows, ignore_index=True)
    print("Generated:", len(new_rows_df), "new transitivity relations")
    return pd.concat([df, new_rows_df], ignore_index=True)



def add_symmetry(df, symmetry_rules, label_mapping, seen_rows):
    # Lọc các hàng có nhãn nằm trong symmetry_rules
    valid_rows = df[df["label"].isin(symmetry_rules)]

    # Áp dụng quy tắc đối xứng để tạo nhãn mới
    valid_rows = valid_rows.assign(
        new_entity1_id=valid_rows["entity2_id"],
        new_entity2_id=valid_rows["entity1_id"],
        new_label=valid_rows["label"].map(symmetry_rules),
        new_entity1_start=valid_rows["entity2_start"],
        new_entity2_start=valid_rows["entity1_start"],
        new_entity1_end=valid_rows["entity2_end"],
        new_entity2_end=valid_rows["entity1_end"],
        new_entity1_text=valid_rows["entity2_text"],
        new_entity2_text=valid_rows["entity1_text"]
    )

    # Tạo hàng mới dưới dạng tuple để kiểm tra trùng lặp
    valid_rows["new_row"] = valid_rows.apply(
        lambda row: (
            row["new_entity1_id"], row["new_entity2_id"], row["document_id"], row["new_label"]
        ),
        axis=1
    )
    # Lọc các hàng chưa có trong seen_rows
    unique_new_rows = valid_rows[~valid_rows["new_row"].isin(seen_rows)]

    # Thêm các hàng mới vào seen_rows
    seen_rows.update(unique_new_rows["new_row"].tolist())

    # Tạo DataFrame các hàng mới
    new_rows = unique_new_rows[[
        "new_entity1_id", "new_entity2_id",
        "new_entity1_start", "new_entity2_start",
        "new_entity1_end", "new_entity2_end",
        "new_entity1_text", "new_entity2_text",
        "document_id", "text", "new_label"
    ]].rename(columns={
        "new_entity1_id": "entity1_id",
        "new_entity2_id": "entity2_id",
        "new_entity1_start": "entity1_start",
        "new_entity2_start": "entity2_start",
        "new_entity1_end": "entity1_end",
        "new_entity2_end": "entity2_end",
        "new_entity1_text": "entity1_text",
        "new_entity2_text": "entity2_text",
        "new_label": "label"
    })

    # Thêm cột label_encoded vào các hàng mới
    new_rows["label_encoded"] = new_rows["label"].map(label_mapping)

    print("Generated:", len(new_rows), "new symmetry relations")
    return pd.concat([df, new_rows], ignore_index=True)
    
def data_augmentation(df, transitivity_rules, symmetry_rules, label_mapping):
    # Tập hợp để lưu các dòng đã thêm (tránh trùng lặp)
    seen_rows = set((row["entity1_id"], row["entity2_id"], row["document_id"], row["label"]) for _, row in df.iterrows())

    previous_len_df = len(df)
    while True:
        df = add_symmetry(df, symmetry_rules, label_mapping, seen_rows)
        df = add_transitivity(df, transitivity_rules, label_mapping, seen_rows)
        current_len_df = len(df)
        if current_len_df != previous_len_df:
            previous_len_df = current_len_df
        else:
            break
    return df
```

```python
def compare_data_distribution(df1, df2):
    # Thống kê số lượng các label
    label_counts_1 = df1["label"].value_counts()
    label_counts_2 = df2["label"].value_counts()
    
    # Tạo phân phối (tỷ lệ phần trăm)
    label_distribution_1 = label_counts_1 / label_counts_1.sum()
    label_distribution_2 = label_counts_2 / label_counts_2.sum()
    
    # Hiển thị thống kê
    print("Number of labels in df1:", label_counts_1.sum())
    print(label_counts_1)
    print("\nNumber of labels in df2:", label_counts_2.sum())
    print(label_counts_2)

    print("\nLabel distribution in df1:")
    print(label_distribution_1)
    print("\nLabel distribution in df2:")
    print(label_distribution_2)
    
    # Vẽ đồ thị biểu diễn phân phối của label
    fig, ax = plt.subplots(1, 2, figsize=(16, 5))

    # Biểu đồ cho df1
    ax[0].bar(label_counts_1.index, label_counts_1.values, color='skyblue', alpha=0.8)
    ax[0].set_xlabel("Labels")
    ax[0].set_ylabel("Number of labels")
    ax[0].set_title("Label distribution in df1")
    ax[0].tick_params(axis='x', rotation=45)

    # Biểu đồ cho df2
    ax[1].bar(label_counts_2.index, label_counts_2.values, color='salmon', alpha=0.8)
    ax[1].set_xlabel("Labels")
    ax[1].set_ylabel("Number of labels")
    ax[1].set_title("Label distribution in df2")
    ax[1].tick_params(axis='x', rotation=45)

    plt.tight_layout()
    plt.show()
```

```python
def show_data_distribution(df):
    # Thống kê số lượng các label
    label_counts = df["label"].value_counts()
    # Tạo phân phối (tỷ lệ phần trăm)
    label_distribution = label_counts / label_counts.sum()
    
    # Hiển thị thống kê và phân phối
    print("Number of label:", label_counts.sum())
    print(label_counts)
    print("\nLabel distribution:")
    print(label_distribution)
    
    # Vẽ đồ thị biểu diễn phân phối của label
    plt.figure(figsize=(8, 5))
    plt.bar(label_counts.index, label_counts.values, color='skyblue', alpha=0.8)
    plt.xlabel("Labels")
    plt.ylabel("Number of label")
    plt.title("Label distribution:")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
# === 2. Dataset Class ===
class TemporalRelationDataset(Dataset):
    def __init__(self, dataframe, tokenizer_spacy, tokenizer_bert, nlp_spacy, bert_model, max_length, pooling_method='mean', window_size=128, device="cpu"):
        self.dataframe = dataframe
        self.tokenizer_spacy = tokenizer_spacy
        self.tokenizer_bert = tokenizer_bert
        self.nlp = nlp_spacy
        self.bert_model = bert_model.to(device)
        self.MAX_LENGTH = max_length
        self.window_size = window_size
        self.pooling_method =  pooling_method
        self.device = torch.device(device)
        
    def __len__(self):
        return len(self.dataframe)
    
    def make_ids_for_bert(self, index):
        # items = []
        # df_length = len(self.dataframe)
        # for index in tqdm(range(df_length), desc="Processing tasks:"):
        row = self.dataframe.iloc[index]
        text = row['text']
        e1_start = row['entity1_start']
        e1_end = row['entity1_end']
        e2_start = row['entity2_start']
        e2_end = row['entity2_end']
        e1_id = row['entity1_id']
        e2_id = row['entity2_id']
        doc_id = row['document_id']
        
        marked_text_e1, marked_text_e2 = self.create_marked_text(text, e1_start, e1_end, e2_start, e2_end)

        # Token hóa
        encoding_e1 = self.tokenizer_bert(marked_text_e1, max_length=self.MAX_LENGTH, padding="max_length", truncation=True, return_tensors="pt")
        encoding_e2 = self.tokenizer_bert(marked_text_e2, max_length=self.MAX_LENGTH, padding="max_length", truncation=True, return_tensors="pt")
        
        input_ids_e1 = encoding_e1['input_ids'].squeeze(0)
        attention_mask_e1 = encoding_e1['attention_mask'].squeeze(0)

        input_ids_e2 = encoding_e2['input_ids'].squeeze(0)
        attention_mask_e2 = encoding_e2['attention_mask'].squeeze(0)

        input_ids = [input_ids_e1, input_ids_e2]
        attention_mask = [attention_mask_e1, attention_mask_e2]
        label = torch.tensor(row['label_encoded'], dtype=torch.long)

        item = {
                'input_ids': input_ids,
                'attention_mask': attention_mask,
                'label': label,
                'tag_id': [e1_id, e2_id, doc_id]
                }

        return item
            # items.append(item)
        # return items

    def get_spans_of_sentences(self, text):
        spans = []
        start = 0
        doc = self.nlp(text)
        sentences = [sent.text.strip() for sent in doc.sents]
        spans = []
        start = 0
        for i, sentence in enumerate(sentences):
            match = re.search(re.escape(sentence), text)
            if match:
                spans.append(match.span())
            else:
                print(f"Not found sentence {sentence}")
        return spans
    
    def find_sentence_containing_span(self, text, spans, target_start, target_end):
        for i, (start, end) in enumerate(spans):
            if start <= target_start < end or start < target_end <= end:
                return text[start:end], (start, end)
        return None, None
    
    def get_token_for_char(self, tokens, char_idx):
        for i, token in enumerate(tokens):
            if char_idx > token.idx:
                continue
            if char_idx == token.idx:
                return i, token
            if char_idx < token.idx:
                return i - 1, tokens[i - 1]
        return len(tokens) - 1, tokens[len(tokens) - 1]
        
    def get_context_by_window(self, text, tokens, e_start, e_end):
        start_token, _ = self.get_token_for_char(tokens, e_start)
        end_token, _ = self.get_token_for_char(tokens, e_end)
        start_token -= (self.window_size - (end_token - start_token)) // 2
        end_token += (self.window_size - (end_token - start_token)) // 2
        end_token += max(0, -start_token)
        start_token = max(0, start_token)
        end_token = min(end_token, len(tokens) - 1)
        start = tokens[start_token].idx
        end = tokens[end_token].idx + len(tokens[end_token])
        return text[start:end], (start, end)
    
    def create_marked_text(self, text, e1_start, e1_end, e2_start, e2_end):
        tokens = self.tokenizer_spacy(text)
        spans = self.get_spans_of_sentences(text)
        context_e1, span_context_e1 = self.find_sentence_containing_span(text, spans, e1_start, e1_end)
        context_e2, span_context_e2 = self.find_sentence_containing_span(text, spans, e2_start, e2_end)
    
        if context_e1 is None or span_context_e1 is None or len(self.tokenizer_spacy(context_e1)) > self.window_size:
            context_e1, span_context_e1 = self.get_context_by_window(text, tokens, e1_start, e1_end)
    
        if context_e2 is None or span_context_e2 is None or len(self.tokenizer_spacy(context_e2)) > self.window_size:
            context_e2, span_context_e2 = self.get_context_by_window(text, tokens, e2_start, e2_end)
        
        e1_start_in_context = e1_start - span_context_e1[0]
        e1_end_in_context = e1_end - span_context_e1[0]
        e2_start_in_context = e2_start - span_context_e2[0]
        e2_end_in_context = e2_end - span_context_e2[0]
    
        marked_text_e1 = (context_e1[:e1_start_in_context] 
                          + "[E1] " 
                          + context_e1[e1_start_in_context:e1_end_in_context] 
                          + " [/E1]" 
                          + context_e1[e1_end_in_context:])
    
        marked_text_e2 = (context_e2[:e2_start_in_context] 
                          + "[E2] " 
                          + context_e2[e2_start_in_context:e2_end_in_context] 
                          + " [/E2]" 
                          + context_e2[e2_end_in_context:])
    
        return marked_text_e1, marked_text_e2
    
    def get_entity_embeddings(self, last_hidden_state, input_ids, start_token, end_token):
        # get position of start_token và end_token
        start_positions = (input_ids == self.tokenizer_bert.convert_tokens_to_ids(start_token)).nonzero(as_tuple=True)
        end_positions = (input_ids == self.tokenizer_bert.convert_tokens_to_ids(end_token)).nonzero(as_tuple=True)
        
        if start_positions[0].size(0) == 0 or end_positions[0].size(0) == 0:
            raise ValueError(f"Token {start_token} or {end_token} not found in input_ids.")
        
        # take embedding from [start_token] to [end_token]
        embeddings = []
        for batch_idx in range(len(start_positions[0])):
            start_idx = start_positions[1][batch_idx]
            end_idx = end_positions[1][batch_idx]
            embeddings.append(last_hidden_state[batch_idx, start_idx:end_idx + 1, :])
        
        return embeddings
    
    def process_entity_embeddings(self, embeddings):
        if self.pooling_method == "mean":
            return torch.stack([torch.mean(embed, dim=0) for embed in embeddings])  # Mean pooling
        elif self.pooling_method == "max":
            return torch.stack([torch.max(embed, dim=0).values for embed in embeddings])  # Max pooling
        elif self.pooling_method == "concat":
            return torch.cat([embed.view(-1) for embed in embeddings], dim=0)  # Concat
        else:
            raise ValueError("Unsupported pooling method. Choose from 'mean', 'max', or 'concat'.")

    def save_to_pickle(self, features, labels, tag_ids, file_name):
        with open(file_name, 'wb') as f:
            pickle.dump({'features': features, 'labels': labels, 'tag_ids':tag_ids}, f)
        print(file_name, "is saved.")
    
    def create_cache_dataset(self, save_path):
        prefix = '/'.join(save_path.split('/')[:-1])
        if os.path.exists(prefix) is False:
            raise ValueError(prefix, "is not existed.")
        
        features = []
        labels =  []
        tag_ids = []
        dataset_length = len(self.dataframe)
        print("Number of item in dataset:", dataset_length)
        print("Make embeddings")
        for i in tqdm(range(dataset_length), desc="Processing tasks:"):
            item = self.make_ids_for_bert(i)
            input_ids = item['input_ids']
            attention_mask = item['attention_mask']
            input_ids_e1 =  input_ids[0].view(1, -1).to(self.device)
            input_ids_e2 =  input_ids[1].view(1, -1).to(self.device)
            attention_mask_e1 =  attention_mask[0].view(1, -1).to(self.device)
            attention_mask_e2 =  attention_mask[1].view(1, -1).to(self.device)
            
            last_hidden_state_e1 = self.bert_model(input_ids=input_ids_e1, attention_mask=attention_mask_e1).last_hidden_state # (batch_size, seq_len, hidden_size)
            last_hidden_state_e2 = self.bert_model(input_ids=input_ids_e2, attention_mask=attention_mask_e2).last_hidden_state # (batch_size, seq_len, hidden_size)
            
            e1_embeddings = self.get_entity_embeddings(last_hidden_state_e1, input_ids_e1, "[E1]", "[/E1]")
            e2_embeddings = self.get_entity_embeddings(last_hidden_state_e2, input_ids_e2, "[E2]", "[/E2]")
            
            # Gộp hoặc tính trung bình embedding
            e1_processed = self.process_entity_embeddings(e1_embeddings)
            e2_processed = self.process_entity_embeddings(e2_embeddings)
            
            # Kết hợp embedding
            x = torch.cat((e1_processed, e2_processed), dim=1).detach().cpu()
            y = item['label'].numpy()
            tag_id = item['tag_id']
            features.append(x)
            labels.append(y)
            tag_ids.append(tag_id)
            
        
        features = np.vstack(features)  # Stack features vertically
        labels = np.hstack(labels)  # Stack labels horizontally
        tag_ids = np.vstack(tag_ids)
        print("Saving dataset")
        self.save_to_pickle(features, labels, tag_ids, file_name=save_path) 
```
