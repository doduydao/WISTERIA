```python
def invert_entry(entry):
    """
    Đảo e1 <-> e2 trong 1 data entry và đổi relation tương ứng
    """

    new_entry = entry.copy()
    
    # Đảo e1/e2 text, span
    new_entry['e1_text'], new_entry['e2_text'] = entry['e2_text'], entry['e1_text']
    new_entry['e1_span'], new_entry['e2_span'] = entry['e2_span'], entry['e1_span']
    
    # Đảo tokens
    new_entry['e1_tokens'], new_entry['e2_tokens'] = entry['e2_tokens'], entry['e1_tokens']
    new_entry['e1_topk'], new_entry['e2_topk'] = entry['e2_topk'], entry['e1_topk']
    new_entry['e1_scores'], new_entry['e2_scores'] = entry['e2_scores'], entry['e1_scores']
    
    # Đảo relation / prediction
    invert_map = {
        "BEFORE": "AFTER",
        "AFTER": "BEFORE",
        "EQUAL": "EQUAL",
        "VAGUE": "VAGUE",
        "OVERLAP": "OVERLAP"
    }
    new_entry['relation'] = invert_map.get(entry['relation'], entry['relation'])
    new_entry['prediction'] = invert_map.get(entry['prediction'], entry['prediction'])
    
    return new_entry
```

```python
def add_inverted_entries(data):
    """
    Thêm các entry đảo ngược vào data, tránh duplicate
    """
    existing_keys = set()
    for entry in data:
        key = (entry['e1_text'], entry['e2_text'], entry['relation'], entry['context'])
        existing_keys.add(key)

    new_entries = []
    for entry in data:
        inverted = invert_entry(entry)
        inv_key = (inverted['e1_text'], inverted['e2_text'], inverted['relation'], inverted['context'])
        if inv_key not in existing_keys:
            new_entries.append(inverted)
            existing_keys.add(inv_key)  # thêm key để tránh trùng entry khác

    data.extend(new_entries)
    return data
```

```python
def create_vocabulary(data, with_topk=False):
    
    pos_set, dep_set, morph_set = set(), set(), set()
    for entry in data:
        if with_topk == True:
            tokens =  entry["e1_tokens"] + entry["e2_tokens"] + entry["e1_topk"] + entry["e2_topk"] + entry["p_topk"]
        else:
            tokens =  entry["e1_tokens"] + entry["e2_tokens"]
        for tok in tokens:
            pos_set.add(tok["pos"])
            dep_set.add(tok["dep"])
            morph_set.add(tok.get("morph", ""))

    pos2id = {pos:i for i, pos in enumerate(sorted(pos_set))}
    dep2id = {dep:i for i, dep in enumerate(sorted(dep_set))}
    morph2id = {m:i for i, m in enumerate(sorted(morph_set))}

    return pos2id, dep2id, morph2id
```

```python
def tokens_to_feature(tokens, pos2id, dep2id, morph2id, max_tokens=None):
    """
    Chuyển token list thành feature vector. 
    Nếu max_tokens có giá trị -> padding/truncate.
    """
    n_tokens = len(tokens) if max_tokens is None else max_tokens
    pos_vec = np.zeros((n_tokens, len(pos2id)))
    dep_vec = np.zeros((n_tokens, len(dep2id)))
    morph_vec = np.zeros((n_tokens, len(morph2id)))
    
    for i, tok in enumerate(tokens[:n_tokens]):
        if tok["pos"] in pos2id:
            pos_vec[i, pos2id[tok["pos"]]] = 1.0
        if tok["dep"] in dep2id:
            dep_vec[i, dep2id[tok["dep"]]] = 1.0
        morph_val = tok.get("morph", "")
        if morph_val in morph2id:
            morph_vec[i, morph2id[morph_val]] = 1.0
            
    return np.concatenate([pos_vec.flatten(), dep_vec.flatten(), morph_vec.flatten()])
```

```python
def entry_to_vector(entry, pos2id, dep2id, morph2id, max_entity_tokens=10, with_topk=False):
    vecs = []
    
    # entity tokens
    vecs.append(tokens_to_feature(entry["e1_tokens"], pos2id, dep2id, morph2id, max_tokens=max_entity_tokens))
    vecs.append(tokens_to_feature(entry["e2_tokens"], pos2id, dep2id, morph2id, max_tokens=max_entity_tokens))

    if with_topk == True:
        # top-k tokens
        for key in ["e1_topk", "e2_topk", "p_topk"]:
            vecs.append(tokens_to_feature(entry[key], pos2id, dep2id, morph2id, max_tokens=None))
        
    return np.concatenate(vecs)
```

```python
def create_feature_names(max_entity_tokens, data, pos2id, dep2id, morph2id, with_topk=False):
    feature_names = []
    
    # 2 entity tokens: e1, e2, max_entity_tokens = 10
    for entity in ["e1", "e2"]:
        for t in range(max_entity_tokens):  # max tokens
            for pos in pos2id:
                feature_names.append(f"{entity}_POS_{pos}")
            for dep in dep2id:
                feature_names.append(f"{entity}_DEP_{dep}")
            for morph in morph2id:
                feature_names.append(f"{entity}_MORPH_{morph}")
    
    # top-k tokens: lấy max k trong dataset
    if with_topk == True:
        k_e1_topk = max(len(entry["e1_topk"]) for entry in data)
        k_e2_topk = max(len(entry["e2_topk"]) for entry in data)
        k_pair = max(len(entry["p_topk"]) for entry in data)
        
        for entity, k in zip(["e1_topk","e2_topk","pair"], [k_e1_topk, k_e2_topk, k_pair]):
            for t in range(k):  # mỗi token top-k
                for pos in pos2id:
                    feature_names.append(f"{entity}_POS_{pos}")
                for dep in dep2id:
                    feature_names.append(f"{entity}_DEP_{dep}")
                for morph in morph2id:
                    feature_names.append(f"{entity}_MORPH_{morph}")
    
    print("Number of feature_names:", len(feature_names))
    return feature_names
```

```python
def create_data_for_DTC(data, with_topk=False, max_entity_tokens=5, is_I2B2=True, do_symmetric=False):
    label2id = {}
    if is_I2B2:
        label2id = {"AFTER":0, "BEFORE":1, "OVERLAP":2}
    id2label = {v:k for k,v in label2id.items()}
    
    if do_symmetric:
        data = add_inverted_entries(data)

    pos2id, dep2id, morph2id = create_vocabulary(data, with_topk=with_topk)
    
    X = np.array([entry_to_vector(entry, pos2id, dep2id, morph2id, max_entity_tokens=max_entity_tokens, with_topk=with_topk) for entry in data])
    y = np.array([label2id[entry["relation"]] for entry in data])
    
    feature_names = create_feature_names(max_entity_tokens, data, pos2id, dep2id, morph2id, with_topk=with_topk)
    
    return X, y, feature_names, pos2id, dep2id, morph2id, label2id, id2label
```

```python
def decode_tree(clf, feature_names, id2label):
    tree_ = clf.tree_
    feature_name = [
        feature_names[i] if i != _tree.TREE_UNDEFINED else "undefined!"
        for i in tree_.feature
    ]
    
    def recurse(node, depth):
        indent = "  " * depth
        if tree_.feature[node] != _tree.TREE_UNDEFINED:
            # One-hot feature => TRUE / FALSE
            name = feature_name[node]
            threshold = tree_.threshold[node]
            if threshold == 0.5:
                # 0 -> FALSE, 1 -> TRUE
                threshold_str = "FALSE / TRUE"
            else:
                threshold_str = str(threshold)
            
            # Nhánh trái: <= threshold → FALSE
            print(f"{indent}if {name} == FALSE:")
            recurse(tree_.children_left[node], depth + 1)
            
            # Nhánh phải: > threshold → TRUE
            print(f"{indent}else:  # {name} == TRUE")
            recurse(tree_.children_right[node], depth + 1)
        else:
            # leaf node
            # decode class
            value = tree_.value[node][0]
            class_idx = value.argmax()
            label = id2label[class_idx]
            print(f"{indent}=> class: {label}")
    
    recurse(0, 0)
```

```python
def decode_tree_with_weights(clf, feature_names, id2label):
    tree_ = clf.tree_
    feature_name = [
        feature_names[i] if i != _tree.TREE_UNDEFINED else "undefined!"
        for i in tree_.feature
    ]
    
    def recurse(node, depth):
        indent = "|   " * depth
        if tree_.feature[node] != _tree.TREE_UNDEFINED:
            name = feature_name[node]
            threshold = tree_.threshold[node]
            n_samples = int(tree_.weighted_n_node_samples[node])
            
            # One-hot feature → TRUE / FALSE
            if threshold == 0.5:
                left_val = "FALSE"
                right_val = "TRUE"
            else:
                left_val = f"<= {threshold:.2f}"
                right_val = f"> {threshold:.2f}"
            
            # Nhánh trái
            print(f"{indent}|--- {name} = {left_val} [samples={n_samples}]")
            recurse(tree_.children_left[node], depth + 1)
            
            # Nhánh phải
            print(f"{indent}|--- {name} = {right_val} [samples={n_samples}]")
            recurse(tree_.children_right[node], depth + 1)
        else:
            # Leaf node
            n_samples = int(tree_.weighted_n_node_samples[node])
            value = tree_.value[node][0]
            class_idx = value.argmax()
            label = id2label[class_idx]
            
            # Xác suất cho mỗi class
            probs = (value / value.sum()) if value.sum() > 0 else value
            probs_str = {id2label[i]: round(float(probs[i]), 4) for i in range(len(probs))}
            
            print(f"{indent}|--- class: {label} [samples={n_samples}, probs={probs_str}]")
    
    recurse(0, 0)
```

```python
def sort_fol_parts(fol_parts):
    """
    fol_parts: list[str], ví dụ: POS(e1, NOUN), ¬DEP(e2, nsubj), ...
    Sắp xếp theo:
        1. entity: e1 trước e2
        2. feature type: POS -> DEP -> MORPH
    """
    def sort_key(pred):
        # Loại bỏ ¬ nếu có
        pred_clean = pred[1:] if pred.startswith("¬") else pred
        
        # Xác định entity
        if pred_clean.startswith("POS("):
            feature_type = 0
        elif pred_clean.startswith("DEP("):
            feature_type = 1
        elif pred_clean.startswith("MORPH("):
            feature_type = 2
        else:
            feature_type = 3  # khác
        
        # Lấy entity
        if "(" in pred_clean:
            entity = pred_clean.split("(")[1].split(",")[0].strip()
        else:
            entity = ""
        
        # key = (entity, feature_type)
        return (entity, feature_type)
    
    return sorted(fol_parts, key=sort_key)
```

```python
def generate_rules_from_tree(clf, feature_names, id2label):
    """
    Trích xuất luật FOL-like từ Decision Tree classifier
    - Kết hợp tất cả feature TRUE/FALSE từ root -> leaf
    - Leaf node in class + tổng samples
    """
    tree_ = clf.tree_
    feature_name = [
        feature_names[i] if i != _tree.TREE_UNDEFINED else "undefined!"
        for i in tree_.feature
    ]
    
    rules = []

    def recurse(node, path_features):
        if tree_.feature[node] != _tree.TREE_UNDEFINED:
            name = feature_name[node]
            threshold = tree_.threshold[node]

            # One-hot threshold → TRUE/FALSE
            if threshold == 0.5:
                left_val = "FALSE"
                right_val = "TRUE"
            else:
                left_val = f"<= {threshold:.2f}"
                right_val = f"> {threshold:.2f}"

            # Nhánh trái
            recurse(tree_.children_left[node], path_features + [(name, left_val)])
            # Nhánh phải
            recurse(tree_.children_right[node], path_features + [(name, right_val)])
        else:
            # Leaf node
            n_samples = int(tree_.weighted_n_node_samples[node])
            value = tree_.value[node][0]
            class_idx = value.argmax()
            label = id2label[class_idx]

            # Build FOL-like rule
            fol_parts = []
            for f, val in path_features:
                if "_POS_" in f:
                    entity, pos = f.split("_POS_")
                    f_str = f"POS({entity}, {pos})"
                elif "_DEP_" in f:
                    entity, dep = f.split("_DEP_")
                    f_str = f"DEP({entity}, {dep})"
                elif "_MORPH_" in f:
                    entity, morph = f.split("_MORPH_")
                    f_str = f"MORPH({entity}, {morph})"
                else:
                    f_str = f

                if val == "TRUE":
                    predicate = f"{f_str}"
                else:
                    predicate = f"\u00AC{f_str}" # \u00AC = ¬
                fol_parts.append(predicate)
                
            fol_rule = " ^ ".join(sort_fol_parts(fol_parts)) + f" = {label}(e1, e2) (samples={n_samples})"
            rules.append(fol_rule)

    recurse(0, [])
    return rules
```

```python
def sort_fol_rules_by_label(fol_rules):
    """
    Sắp xếp danh sách luật FOL theo label và sau đó theo số samples giảm dần.
    fol_rules: list[str], mỗi luật có dạng
    "POS(e1, NOUN) ^ DEP(e1, nsubj) = BEFORE (samples=123)"
    """
    def sort_key(rule):
        try:
            # Lấy label
            label_part = rule.split(" = ")[-1]
            label = label_part.split("(")[0].strip()
            
            # Lấy số sample
            samples_str = label_part.split("samples=")[-1].split(")")[0]
            samples = int(samples_str)
            
            # key: label alphabetically, samples descending
            return (label, -samples)
        except:
            return ("", 0)
    
    return sorted(fol_rules, key=sort_key)
```

```python
def group_fol_rules_by_label(fol_rules, with_clean_rule=True):
    """
    fol_rules: list of strings, mỗi luật có dạng
    "POS(e1, NOUN) ^ DEP(e1, nsubj) = BEFORE (samples=123)"
    
    Trả về dict: {label: [rule1, rule2, ...]}
    """
    rule_dict = defaultdict(list)
    
    for rule in fol_rules:
        # tách label nằm trước "(samples="
        try:
            label_part = rule.split(" = ")[-1]  # phần cuối
            label = label_part.split("(")[0].strip()  # BEFORE / AFTER / OVERLAP ...
            if with_clean_rule:
                rule = rule.split(" (samples=")[0]
            rule_dict[label].append(rule)
        except:
            continue  # nếu lỗi skip
    
    return dict(rule_dict)  # trả về dict bình thường

```
